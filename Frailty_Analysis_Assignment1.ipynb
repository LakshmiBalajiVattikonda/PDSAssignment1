## **1. Frailty Analysis**



```python
import pandas as pd
import numpy as np
```


* pandas libarary is used for the data manipulation and analysis.
* numpy libarary is used for the mathematical operations.




### **Stage 1: Ingest (Data input)**

```python
df = pd.read_csv("Frailty Analysis data.csv")
print("Raw Data:")
display(df)
```

### **Stage 2: Process (Data Processing)**

**(a) Unit Standardization:**


First, height is converted from inches to meters using the formula: inches * 0.0254.


Next, weight is converted from lb to kilograms using the formula: lb * 0.45359237

```python
df["Height_m"] = df["Height_in"] * 0.0254
df["Weight_kg"] = df["Weight_lb"] * 0.45359237
display(df)
```

**(b) Feature Engineering:**


BMI is calculated using the formula: weight (kg) / height (m)².


The age groups are separated into four categories: <30, 30–45, 46–60, and >60.


```python
df["BMI"] = (df["Weight_kg"] / (df["Height_m"]**2)).round(2)
display (df)
```

```python
def age_group(age):
    if age < 30:
        return "<30"
    elif 30 <= age <= 45:
        return "30-45"
    elif 46 <= age <= 60:
        return "46-60"
    else:
        return ">60"

df["AgeGroup"] = df["Age"].apply(age_group)
display(df)
```

**(c) Categorical → Numeric Encoding:**

By using the map function, 'N' is converted to 0 and 'Y' is converted to 1

int8 makes the column memory-efficient


```python
df["Frailty_binary"] = df["Frailty "].str.strip().map({"N":0, "Y":1}).astype("int8")
display(df)
```

**One-hot encoding AgeGroup**

Converts the AgeGroup column into separate dummy variables:

AgeGroup_<30, AgeGroup_30-45, AgeGroup_46-60, AgeGroup_>60 and If a person belongs to a particular age group, the value is 1; otherwise, it is 0.



```python
df["AgeGroup <30"] = np.where(df["AgeGroup"] == '<30', 1, 0)
df["AgeGroup 30_45"] = np.where(df["AgeGroup"] == '30-45', 1, 0)
df["AgeGroup 40_60"] = np.where(df["AgeGroup"] == '45-60', 1, 0)
df["AgeGroup >60"] = np.where(df["AgeGroup"] == '>60', 1, 0)
display(df)
```

### **Stage 3: Analyze (Data Analysis)**

**(d) EDA & Reporting**

i) Summary statistics:

*  df.describe() generates statistics for each numeric column.
*  ".T" transposes rows/columns for better readability.
*  We select mean, std, min, and max values.
* This helps understand the distribution of data








```python
summary_stats = df.describe().T[["mean","std","min","max"]]
display(summary_stats)
```

ii) Correlation (Grip strength vs Frailty)


*  Calculates Pearson correlation between grip strength and frailty.
*  A negative correlation (like -0.71) means:
    *  Higher grip strength → Lower frailty risk.








```python
correlation = df["Grip_strength"].corr(df["Frailty_binary"])
print("\nCorrelation (Grip_strength vs Frailty_binary):", correlation)
```
